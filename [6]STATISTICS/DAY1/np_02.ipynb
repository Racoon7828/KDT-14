## [ Numpy - ndarray 원소 접근] 
<hr>

- 인덱싱 기반 : 왼 -> 오, 오 -> 왼 2개 종류 인덱스
- 슬라이싱 : 연속적이거나 일정한 규칙이 있는 원소 접근 방식

```python
# 모듈 로딩
import numpy as np
import pandas as pd
```

```python
# 실습 데이터 생성
data = np.arange(2, 20, 3)
data
```

```python
# ndarray 원소 접근 - 인덱싱
print(data[0], type(data[0]), data[0].ndim)
print(data[-1], type(data[-1]), data[-1].ndim)
```

```python
# 원소값 변경
data[0] = 99
print(data)
```

```python
# ndarray 원소 접근 - 슬라이싱
data[0:-2]
```

```python
# 슬라이싱 + 값 변경
# View 타입 : 값 변경 시 원본 데이터 변경 됨
data[0:-3] = 7

# copy 타입 : 복사본 즉, 메모리 새롭게 저장, 값 변경 시 원본 데이터 변경 X
data2 = data[0:3].copy()
data2[0:-1] = 9

print(data, data2, sep='\n')
```

### [ Numpy Fancy Indexing ] <hr>
- 원하는 데이터의 인덱스를 배열/리스트에 담아서 한꺼번에 추출하는 인덱싱
- 예 : 배열명[[인덱스, 인덱스]]

```python
import numpy as np
```

```python
dataNP = np.array([[11,22,33], [44,55,66]], dtype=np.int8)
```

```python
# 팬시 인덱싱 => 2번, 0번 컬럼 추출 
dataNP[:, [2,0]] = 9
dataNP
```

```python
# 팬시 인덱싱 => 1번, 0번 컬럼 추출 
dataNP2 = dataNP[[1,0]]
dataNP2
```
