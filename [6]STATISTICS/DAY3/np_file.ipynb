## [ Numpy 파일 입출력 ] <hr>

- ndarray 객체를 파일로 저장
- 확장자 : 1개 파일 -> 파일명.npy / 여러개 파일 -> 파일명.npz

```python
import numpy as np
```

```python
# 실습 array 생성
a1 = np.linspace(1, 3)
a2 = np.linspace(3, 5)
```

```python
# ndarray 저장
data_file = './data.npy'
all_file = './all.npz'
com_file = './com.npz'
txt_file = './text.out'

# 1개 ndarray => 파일로 저장
np.save(data_file, a1)

# 여러개 ndarray => 파일로 저장
np.savez(all_file, a1, a2, a1)

# 여러개 ndarray => 압축 후 파일로 저장
np.savez_compressed(com_file, x=a1, y=a2, z=a1)
                    
# 여러개 ndarray => 텍스트 파일로 저장
np.savetxt(txt_file, a1)
```

### ndarray 읽기 
<hr>

```python
# 파일 1개 읽기
np.load('data.npy')
```

```python
# 여러개 array 저장된 파일 읽기
ret = np.load('all.npz')
print(f'ret => {type(ret)}')
print(f'files => {ret.files}')
print(f'keys => {ret.keys}')
print(f'{ret["arr_0"]}')
```

```python
# 텍스트로 저장된 파일 읽기
np.loadtxt('text.out')
```

```python
# 다양한 텍스트 파일로 저장
arr = np.linspace(1,5,100).reshape(-1,5)

csv_file = './data.csv'
np.savetxt(csv_file,
           arr,
           delimiter=':',
           fmt='%.3f',
           header='[수신 데이터]',
           footer='[--KNU--]')
```
