### [ ndarray 이미지 데이터 ] <hr>
- python에서 이미지 파일 로딩 ==> ndarray 타입
- 화면 그림 출력 : matplotlib.plot.imshow()
- 컴퓨터 비젼처리 : opencv 패키지 - numpy | pillow 패키지 - tensor

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.cm as cm
```

```python
data = np.arange(0, 256).reshape(16,16)
# data
```

```python
plt.imshow(data, cmap=cm.gray)
plt.show()
```

```python
# 임의의 값 추출 => 0 ~ 255 범위에서 랜덤 숫자
# np.random.rand() : 0 <= ~ < 1 
data = np.random.rand(16,16)

# np.random.randint(low, high)
# data = np.random.randint(0,255)

plt.imshow(data, cmap=cm.turbo)
plt.show()
```
