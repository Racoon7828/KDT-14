### [ 유니버설 함수 - Numpy 성능 핵심요소 ] <hr>
- 반복문 없이 원소 단위 연산 수행
- 자동 타입과 형태 변화 최적화 진행 => 브로드캐스팅 자동 진행

```python
import numpy as np
```

```python
arr = np.array([3.2, 0.1, 8., -5.2, -3.12, -1.7])
```

```python
# 실수 데이터 => 소수점 이하 자릿수 관련 함수
# => np.ceil() 소수점 이하 올림 처리
print(f'원본 : {arr}')
print(f'자릿수 처리 (올림) : {np.ceil(arr)}')
```

```python
# => np.round() 소수점 이하 반올림 처리
print(f'원본 : {arr}')
print(f'자릿수 처리 (반올림) : {np.round(arr)}')
```

```python
# => np.floor() 소수점 이하 내림 처리
print(f'원본 : {arr}')
print(f'자릿수 처리 (내림) : {np.floor(arr)}')
```

```python
# 누적 연산 처리 함수
arr = np.arange(1, 10, 3)

# => 연산.reduce(array_obj) : 누적 연산 결과 반환
result1 = np.add.reduce(arr)

result2 = np.multiply.reduce(arr)

print(f'{arr} => {result1}')
print(f'{arr} => {result2}')
```

```python
# 누적 연산 처리 함수
arr = np.arange(1, 17).reshape(4,4)

# => 2개 컬럼만 묶어서 연산
retarr = np.add.reduceat(arr, [0,2])

print(f'{arr} \n --연산후-- \n {retarr}')
```
