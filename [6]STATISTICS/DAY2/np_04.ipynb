## [다양한 ndarray 객체 생성] <hr>

```python
import numpy as np
```

```python
# 다양한 타입의 데이터 => ndarray 저장
# 사용함수 : np.array()

arr1 = np.array([11,22,33])
arr2 = np.array([2.3,0.9])
arr3 = np.array(['A','B'])
arr4 = np.array(['Good',100,True]) # 서로 다른 타입의 데이터의 경우 제일 큰 데이터 타입으로 일치 시킴
                                   # 배열은 동일 자료형만 가능하기 때문
```

```python
# 함수 기능 : ndarray 객체 속성 출력 함수
def print_info(arr, name):
    print(f'\n-- {name} 객체 정보 --')
    print(f'ndim     : {arr.ndim}D')
    print(f'shape    : {arr.shape}')
    print(f'size     : {arr.size}개')
    print(f'dtype    : {arr.dtype}')
    print(f'itemsize : {arr.itemsize}byte')
    print(f'value    : {arr}')
```

```python
# ndarray 객체들 속성들 확인
print_info(arr1, 'arr1')
print_info(arr2, 'arr2')
print_info(arr3, 'arr3')
print_info(arr4, 'arr4')
```
