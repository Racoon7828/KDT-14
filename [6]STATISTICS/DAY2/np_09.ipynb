## [ 축(axis) 변경] <hr>
- shape 속성에 튜플 값의 순서를 변경
    * 2D의 shape : (행,열)
    * 3D의 shape : (깊이, 행, 열)
    * 4D의 shape : (깊이/개수, 깊이, 행, 열)
- 원소 개수 변경 없음, 차원 변경 없음
- 속성 : ndarray.T
- 메서드 : ndarray.transpose(*축들) / ndarray.swapaxes(축1, 축2)

```python
import numpy as np
```

```python
a1 = np.arange(1, 9).reshape(2,4) # 2D
a2 = np.arange(1, 5, 0.25).reshape(1,2,8,1) # 4D

print(a1.shape, a2.shape)
```

```python
# 축 변경 속성 ndarray.T
a1T = a1.T
a2T = a2.T

print(a1.shape, a1T.shape, sep='\n\n')
print(a2.shape, a2T.shape, sep='\n\n')
```

```python
# 축 변경 속성 ndarray.transpose(*axes) : 축 순서 설정
a2T = a2.transpose(1, 0 , 3, 2) # 축 번호 (1,2,8,1)

print(a2.shape, a2T.shape, sep='\n\n')
```

```python
# 축 변경 속성 ndarray.swapaxes(axis1, axis2) : 2개 축 변경
a2T = a2.swapaxes(1, 0)

print(a2.shape, a2T.shape, sep='\n\n')
```
