## [array 차원/형태 변경] <hr>
- N 차원 < --- > 1차원
    * 원소 개수 동일해야 함
    * 함 수 : np.reshape()
    * 메서드 : ndarray.reshape()

```python
import numpy as np
```

```python
arr1d = np.arange(1, 31)
arr2d = arr1d.reshape(-1, 3) # 컬럼 수 3개, 행 수는 자동
arr3d = arr1d.reshape(5, 2, -1) # 깊이/개수 5개, 행 수 2개, 컬럼 자동

print(arr2d, arr3d, sep='\n\n')
```

```python
# reshape()으로 생성된 ndarray 특징
arr2d[0,0] = 100
arr1d[-1] = 77

# => 원본 arr1d 과 변환된 arr2d
print(arr1d, arr2d, sep='\n\n')
```

```python
# Nckdnjs ---> 1차원으로 변환 : 평탄화
# => reshape(-1)
arr3d.reshape(-1)

np.reshape(arr3d, -1)
```

```python
# => flatten()
f = arr3d.flatten()

# arr3d[0,0,1] = 200
f[1] = 200

print(arr3d, f, sep='\n\n')
```

```python
# => ravel()
f = arr3d.ravel()

f[2] = 300

print(arr3d, f, sep='\n\n')
```
