## [ 정렬 & arg 함수명() ] <hr>
- 함수 : np.sort() => [기] 오름차순 정렬된 ndarray 반환
- 메서드 : ndarray.sort() => 원본 정렬됨

- argOOO() 함수들
    - 해당 값이 있는 인덱스를 반환
    - 예 : ndarray.max() - 원소 중 가장 큰 값, ndarray.argmax() : 원소 중 가장 큰 값의 인덱스

```python
import numpy as np
```

```python
arr = np.array([10, 20, 30, 40, 50])
print(arr[0], arr[-1], arr[1:4])
```

```python
a1 = np.array([[3,1,6],
               [0,7,4],
               [5,3,2]])
```

```python
# 정렬
## 함수 : np.sort()
as1 = np.sort(a1, axis=None)
print("정렬 전 ==>", a1.shape , a1 )
print("정렬 후 ==>\n", as1, end='\n\n')

as1 = np.sort(a1, axis=0)
print("행 기준 정렬 후 ==>\n",  as1, end='\n\n')

as1 = np.sort(a1, axis=1)
print("열 기준 정렬 후 ==>\n",  as1, end='\n\n')
```

```python
## 메서드 : ndarray.sort() 원본 변경됨 
##         [기] axis=-1  : -1번 축 기준 오름차순 정렬 
print("정렬 전 ==>", a1 )

## -1번 축 기준 오름차순 정렬 즉, 행방향 정렬
a1.sort()
print("a1  ==>", a1 )

## 열방향 정렬
a1.sort(axis=0)
print("a1  ==>", a1 )
```

```python
# argsort() : 정렬 전 데이터가 있는 인덱스 반환
a1 = np.array([ [3,	1,	6],	
		        [0,	7,	4],	
		        [5,	3,	2]])

arr_idx = np.argsort(a1, axis=1)
print("정렬 전 인덱스 arr_idx : \n", arr_idx)

# 다차원 인덱스를 적용해 원래 배열을 정렬 시 : np.take_along_axis() 함수
sorted_a1 = np.take_along_axis(a1, arr_idx, axis=1)
print("정렬 후 : \n", sorted_a1)
```

## [array 필터링] <hr>
- BlooeanIndexing => 조건에 맞는 데이터/원소 추출
- 함수 : np.where(array[조건식]) => 조건에 맞는 데이터/원소의 인덱스 튜플로 반환

```python
import numpy as np
```

```python
arr1d = np.arange(1, 31)
```

```python
# 필터링 - (1) Boolen Indexing
bool_idx = arr1d % 2 == 0
arr1d[bool_idx]
```

```python
# 필터링 - (2) np.where() : 필터링 된 인덱스
idx = np.where(~arr1d % 2)
arr1d[idx]
```

```python
# 필터링 - (3) np.all() : 필터링 된 인덱스
arr2d = arr1d.reshape(-1, 10)
print(arr2d)

print(f'axis = None : {np.all(arr2d % 2)} / {np.any(arr2d % 2)}')

print(f'axis = 1 : {np.all(arr2d % 2, axis=1)} / {np.any(arr2d % 2, axis=1)}')

print(f'axis = 0 : {np.all(arr2d % 2, axis=0)} / {np.any(arr2d % 2, axis=0)}')
```
