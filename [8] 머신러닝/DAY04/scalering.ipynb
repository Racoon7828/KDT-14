### [스케일러 이해하기] <hr>
- 모델 학습 전에 수치 데이터의 값의 범위를 맞추는 전처리 과정
- 컬럼단위
    * 피쳐/특성별로 크기, 분포를 조정
    * 학습 시 특정 피쳐가 많은 영향을 끼치는 것 막기
- 로우단위

```python
from sklearn.preprocessing import StandardScaler, MinMaxScaler, RobustScaler, MaxAbsScaler
import pandas as pd
```

```python
data = pd.DataFrame({'h':[158, 192, 172, 164, 155],
                     'w':[50, 81, 88, 41, 52]})
```

**scikit-learn 스케일러 사용법**
- 인스턴스 생성
- fit()메서드에 데이터 전달 => 스케일링을 위한 데이터 기반 기준값 계산
- transform()메서드에 데이터 전달 => 데이터 스케일링 진행
- fit_transform() 메서드 => fit() 기능 + transform() 기능

```python
# [1] StandardScaler : 평균 0, 표준편차 1이 되도록 스케일링
# 인스턴스 생성
stdScaler = StandardScaler()

stdScaler.fit(data)

scaled_data = stdScaler.transform(data)

print(scaled_data.mean().round(1), scaled_data.std().round(1))
print(scaled_data)
```

```python
# [2] MinMaxScaler : 0 ~ 1 사이로 변환
# 인스턴스 생성
mmScaler = MinMaxScaler()

mmScaler.fit(data)

scaled_data = mmScaler.transform(data)

print(scaled_data.mean().round(1), scaled_data.std().round(1))
print(scaled_data)
```

```python
# [2] MinMaxScaler : 0 ~ 1 사이로 변환
# 인스턴스 생성
rbScaler = RobustScaler()

rbScaler.fit(data)

scaled_data = rbScaler.transform(data)

print(scaled_data.mean().round(1), scaled_data.std().round(1))
print(scaled_data)
```

```python
# [4] MaxAbsScaler : 최대값과 절대값
maScaler = MaxAbsScaler()

maScaler.fit(data)

scaled_data = maScaler.transform(data)

print(scaled_data.mean().round(1), scaled_data.std().round(1))
print(scaled_data)
```
