```python
from sklearn.preprocessing import LabelEncoder, OneHotEncoder, OrdinalEncoder
import pandas as pd
```

```python
data = pd.DataFrame({'H':[158,192,172,164,155],
                    "W":[50,81,110,41,81],
                    'Blood' : ['B', 'O', 'O', 'A', 'AB'],
                    'Grade' : ['1','2','1','3','2'],
                    'B_TYPE': ['정상', '마름', '비만', '정상', '비만']})
```

```python
# 혈액형 자료형 변환
data['Blood'] = data['Blood'].astype('category')
data['B_TYPE'] = data['B_TYPE'].astype('category')
# data['Grade'] = data['Grade'].astype('category')
```

**인코딩(Encoding)**

```python
# [1] LabelEncoder : 타겟 컬럼의 수치화
#                    .fit(1D)
lblEncoder = LabelEncoder()

lblEncoder.fit(data['B_TYPE'])
print('classes', lblEncoder.classes_)

encode_target = lblEncoder.transform(data['B_TYPE'])

print(data['B_TYPE'].to_list())
print(encode_target)
```

```python
# [2] OneHotEncoder : 텍스트 피챠
ohEncoder = OneHotEncoder(sparse_output=False)

ohEncoder.fit(data[['Blood']])
print(ohEncoder.categories_)
print(ohEncoder.drop_idx_)
print(ohEncoder.feature_names_in_)

encode_df = ohEncoder.transform(data[['Blood']])

print(data['Blood'].to_list())
print(encode_df)
```

```python
# [3] OrdinalEncoder : 텍스트 피쳐. 순서 있는 텍스트 데이터
#                      .fit(2D)
odEncoder = OrdinalEncoder()

odEncoder.fit(data[['Grade']])
print(odEncoder.categories_)
print(odEncoder.feature_names_in_)

encode_df = odEncoder.transform(data[['Grade']])

print(data['Grade'].to_list())
print(encode_df.tolist())
```
