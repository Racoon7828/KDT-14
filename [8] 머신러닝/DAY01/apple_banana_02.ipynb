- **전처리 방법**

````python
import cv2, os
import matplotlib.pyplot as plt
import koreanize_matplotlib
````

````python
# => 이미지 관련
ROI_DIR = '../Data/Roi'

# 수집 데이터 관련
DATA_LABEL = ['apple', 'banana']

# 경로
APPLE_ROI = f'{ROI_DIR}/apple'

BANANA_ROI = f'{ROI_DIR}/banana'

DATA_DIRS = [APPLE_ROI, BANANA_ROI]
````

### [1]데이터 전처리: 크기 일치

````python
# 크기 : 100 x 100
# -> 상대 크기 50% 줄이기
# -> 절대크기 100 x 100
# 이미지의 크기 | 50% 줄인크기
for target in DATA_DIRS:
    filelist = os.listdir(target)
    for filename in filelist:
        IMG_PATH = f'{target}/{filename}'
        SAVE_PATH = f'{target}/re_{filename}'
        img = cv2.imread(IMG_PATH, cv2.IMREAD_GRAYSCALE)
        
        # 비율유지 50% 줄이기
        res = cv2.resize(img, (0,0), fx=0.5, fy=0.5)
        # 절대크기 줄이기
        res = cv2.resize(res, (100,100))

        # 이미지 파일 저장
        cv2.imwrite(SAVE_PATH, res)
````

### [2] 데이터셋 생성
- DataFrame으로 저장하는 방식
- Dataframe => csv 형식 파일 저장
- 형식
    * 픽셀, 픽셀,...,픽셀, 라벨
    * 라벨, 파일명, 픽셀,...,픽셀

- 픽셀 데이터 : cv2.imread() 2D/3D ==> 1D
- 라벨 데이터 : 폴더명 ==> 경로 추출 또는 하드 코딩 지정

````python
import numpy as np
import pandas as pd

data_list = []
for target, label in zip(DATA_DIRS, DATA_LABEL):
    filelist = os.listdir(target)

    for filename in filelist:
        if not filename.startswith("re_"): continue

        IMG_PATH = f'{target}/{filename}'
        img = cv2.imread(IMG_PATH, cv2.IMREAD_GRAYSCALE)
        
        flatten = img.flatten()

        row = flatten.tolist() + [label]
        data_list.append(row)
print(f'data_list : {len(data_list)}')
````

````python
columns = [f'pixel{i}' for i in range(100*100)] + ['label']

SAVE_CSV_PATH = 'fruit_data.csv'

df = pd.DataFrame(data_list, columns=columns)
df.to_csv(SAVE_CSV_PATH, index=False)

print(df.shape)
print(f'저장완료: {SAVE_CSV_PATH}')
````
