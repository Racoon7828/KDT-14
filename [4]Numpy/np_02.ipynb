```python
import numpy as np, matplotlib.pyplot as plt
# cv2 - openCV = (open computer vision library)
import cv2
```

```python
# 영상(image) 정보란 -> (큰)행렬
# gray(1채널) - color(3채널) - RGB
# 3채널의 경우
# (H,W,3)
# dtype - uint8(0~255)

# 1채널
# (H, W)
# dtype - float32/64 (0~1)
# heatmap -> 전형적인 1채널 이미지
```

```python
# 0 행렬 생성
a = np.zeros((5,5), dtype=np.float32)
a = np.arange(25).reshape(5,5)
plt.imshow(a, cmap='magma'); # imshow에서 1채널은 minmax scaling
```

```python
# 컬러 영상 H, W, 3(RGB or BGR)
image = np.zeros((5,5,3), dtype=np.uint8)
image[1,3,0] = 255
image[2,3,0] = 255
image[2,4,0] = 255
image[3,4,0] = 255

# 한줄 파란색
image[-1,:,2] = 255 # magenta(R+B) / cyan(B+G) yellow (R+G)

# 흰색
image[3,1,:] = 255
image[3,2,:] = 255
image[2,2,:] = 255
image[3,3,:] = 255

# 핑크색
image[-1,0,0] = 255
image[-1,0,1] = 150
image[-1,0,2] = 220
image[-2,0,:] = [255,150,220]
image[2,0,:] = [255,150,220]
image[2,1,:] = [255,150,220]

# 노란색
image[:2,:] = [255,255,0]

plt.imshow(image);
```

```python
# 0~1 사이의 값으로 변경 - minmax(0,1)
# minmax scaling -> [0,1]
a = np.array([10,3,2,9,6,4,7,5])
mx = a.max()
mn = a.min()
T = 1
B = -1

# B ~ T
s = (T-B) / (mx - mn)
b = B - s * mn
mn * s + b # = B
```

```python
x = np.array([1,3,5,7])
y = np.array([0,1,2,3])
s = 1/2

# x[0] * s + b = y[0]
b = -1/2

x * s + b # = y
```
