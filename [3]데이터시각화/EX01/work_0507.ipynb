```python
import pandas as pd
df = pd.read_csv(r"C:\Users\Win11Pro\Desktop\KDT-14\[3]데이터시각화\DATA\bike_rentals.csv")
df.info()
```

```python
# int 타입만 선택
df.select_dtypes(include='int')
```

```python
# int 타입 제외 선택
df.select_dtypes(exclude='int')
```

```python
# 새로운 인덱스 datetime(00:00:00) 추가
df = df.set_index('datetime')
df
```

```python
# filter()으로 '00:00:00' 행만 출력
df.filter(like='00:00:00', axis=0)
```

```python
# 'humidity', 'windspeed' 행만 출력
df.filter(items=['humidity', 'windspeed'])
```

```python
# 정규 표현식 사용 = regex=''
# 'in.s' => in 과 s 사이의 임의의 한문자를 지니는 열 => w'inds'peed
# .은 숫자와 알파벳 등을 포함한 임의의 문자 1개
df.filter(regex='in.s')
```

```python
# 행 이름 변경
df = df.rename(
    {'registered':'registered_user',
     'casual':'unregistered_user'},
     axis=1
)
# df.rename(
#     columns = {'registered':'registered_user',
#      'casual':'unregistered_user'}
# )
df
```
