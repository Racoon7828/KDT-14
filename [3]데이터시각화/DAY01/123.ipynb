```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
plt.title('plotting') # 그래프 제목

# 그래프 그리기
plt.plot([10,20,30,40], label='asc', color='red')
plt.plot([40,30,20,10], label='desc', color='yellow')
plt.legend(loc=1) # legend = 범례 표시, loc = 범례 위치
plt.show()
```

```python
# 그래프 선 모양
plt.title('linestyle') # 그래프 제목

# 그래프 그리기
plt.plot([10,20,30,40], label='asc', color='red', linestyle='--')
plt.plot([40,30,20,10], label='desc', color='b', ls=':')
plt.legend(loc=1) # legend = 범례 표시, loc = 범례 위치
plt.show()
```

```python
# 마커 모양 바꾸기
plt.title('marker') # 그래프 제목

# 그래프 그리기
plt.plot([10,20,30,40], 'r.', label='circle')
plt.plot([40,30,20,10], 'g^', label='taiangle')
plt.legend(loc=1) # legend = 범례 표시, loc = 범례 위치
plt.show()
```
