```python
import pandas as pd
import matplotlib.pyplot as plt
```

그래프의 구성요소
- Figure: 그림 전체
- Axes: 그림 내부의 좌표축을 의미하며, 개별 그림을 의미
- Line: 선 그래프에서 선
- Markers: 점 그래프에서 마커
- Legend: 범례
- Title: 제목
- Grid: 격자
- Spines: 윤곽선
- X axis label: x축 레이블
- Y axis label: y축 레이블
- Major tick: 메인 눈금
- Major tick label: 메인 눈금 레이블
- Minor tick: 서브 눈금
- Minor tick label: 서브 눈금 레이블

### 그래프 종류
- plot(선)
- bar(수직 막대)
- barh(수평 막대)
- box(박스 플롯)
- kde(커널 밀도 그래프)
- area(면적 그래프)
- pie(파이 그래프)
- hexbin(고밀도 산점도 그래프)
- scatter(산점도)
- hist(히스토그램)
- sns.heatmap(히트맵)

```python
# 기초 선 그래프 만들기
x = [1,2,3]
y1 = [3,5,1]
y2 = [4,8,4]
plt.plot(x, y1, '--o') # '--' = 점선 / 'o' = 원으로 변곡점 생성
plt.plot(x, y2, '--o')
```

```python
plt.plot(x, y1, '--o')

# plt.show == 현재까지 제작된 그래프를 차트에 생성하고 출력을 마무리 한다.
plt.show()

plt.plot(x, y2, '--o')

# 일반적인 .py 파일에선 새 창에서 그래프를 그린다.
plt.show()
```

```python
# plt 옵션
# 1. figure(figsize=(x,y))
# 전체 그래프 사이즈 조절
plt.figure(figsize=(5,3))
plt.plot(x, y1, '--o')
plt.plot(x, y2, '--o')
plt.show()
```

```python
# 2. xlabel(), ylabel()
# x축, y축 라벨 지정
plt.xlabel('X-VALUE')
plt.ylabel('Y-VALUE')

# 그래프 제목 생성
plt.title('TEST CHART')

plt.plot(x, y1, '--o')
plt.plot(x, y2, '--o')
plt.show()
```

```python
# 3. plt.axis()
# 중심축 설정 ([x시작, x끝 ,y시작 ,y끝])
plt.axis([0,5,0,10])
plt.plot(x, y1, '--o')
```

```python
# 그래프 화면 나누기
# .subplot(행x, 열y, 서브플롯)
# 4개 서브 플롯 = [[1][2]
#                [3][4]]

# 그래프 크기 설정
plt.figure(figsize=(8,5))

# 첫번째 그래프
plt.subplot(2,2,1)
plt.plot(x, y1, '--o')

# 두번째 그래프
plt.subplot(2,2,3)
plt.axis([0,5,0,10])
plt.plot(x, y1, '--o')

# 세번째 그래프
plt.subplot(1,2,2)
plt.plot(x, y2, '--o')
plt.show()

```

```python
# 바 그래프 그리기
x = [1,2,3]
y1 = [3,5,1]
y2 = [4,8,4]
plt.figure(figsize=(5,5))
plt.bar(x,y1,width=0.5)
plt.show()
```

```python
import numpy as np
y3 = np.array(y1) + np.array(y2)

# 바 그래프 곂쳐 그리기
plt.figure(figsize=(5,5))
plt.bar(x,y3,width=0.5)
plt.bar(x,y1,width=0.5)
```

```python
# 파이(원)그래프 그리기
plt.pie(y1, labels=['one','two','three'])
plt.show()
```
