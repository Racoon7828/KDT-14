### [ 이진화 / 쓰레시 홀딩 ] <hr>
- 이미지의 형태 / 회형 . 윤곽 정보를 추출할 수 있는 방식
- 이미지에서 원하는 부분만 추출할 수 있는 방식
- 픽셀 값을 0과 1(또는 255, 특정값)으로 변환하는 것
- 픽셀 값 변환을 위한 기준값/경계값/임계값 설정이 가장 중요

```python
import numpy as np, cv2, matplotlib.pyplot as plt, koreanize_matplotlib
import sys
sys.path.append(r'C:\Users\Win11Pro\Desktop\KDT-14\[7]_CVISION')
from cv_utils import printPolts
```

```python
# 데이터 준비
IMG_FILE = '../Data/images/gray_gradient.jpg'
```

```python
# 이미지 데이터 로딩 및 기본 정보 확인
img = cv2.imread(IMG_FILE, cv2.IMREAD_GRAYSCALE)
print(f'[기본정보] shape : {img.shape}, ndim : {img.ndim}, dtype : {img.dtype}')
```

```python
# 이진화 - 방식 [1]
# 이진화 적용 ndarray 객체 생성
bin = np.zeros_like(img)
print(f'[기본정보] shape : {bin.shape}, ndim : {bin.ndim}, dtype : {bin.dtype}')

# 임계값/경계값/쓰레시홀드
TH_VALUE = 127

# 임계값 보다 큰 픽셀은 255로 값 저장
bin[img > 127] = 255

# 시각화 : 원본, 이진화 => 1행 2열
printPolts(1 ,2 ,['원본이미지', '이진화 이미지'], [img, bin])
printPolts(1 ,2 ,['원본이미지', '이진화 이미지'], [img, bin], ['gray'])

# fig, axes = plt.subplots(1,2,sharey = True)
# for i in range(2):
#     axes[i].imshow([img, bin][i], cmap='gray')
#     axes[i].set_title(['원본이미지', '이진화 이미지'][i])
#     axes[i].axis('off')
```

```python
# # 이진화 - opencv 방식
# 이진화 이미지 생성
TH_VALUE, MAX_VALUE = 127, 255

ret, bun = cv2.threshold(img, TH_VALUE, MAX_VALUE, cv2.THRESH_BINARY)

# 시각화 : 원본, 이진화 => 1행 2열
printPolts(1 ,2 ,['원본이미지', '이진화 이미지'], [img, bin])
printPolts(1 ,2 ,['원본이미지', '이진화 이미지'], [img, bin], ['gray'])

# fig, axes = plt.subplots(1,2,sharey = True)
# for i in range(2):
#     axes[i].imshow([img, bin][i], cmap='gray')
#     axes[i].set_title(['원본이미지', '이진화 이미지'][i])
#     axes[i].axis('off')
```
