### [ 이미지/영상 연산 - 비트 연산 ] <hr>
- 픽셀의 비트단위 연산 수행
- 특정 부분만 추출하거나 사용하기위한 용도
- 종류
    * bitwise_and()
    * bitwise_or()
    * bitwise_xor()
    * bitwise_not()

```python
import numpy as np, cv2, matplotlib.pyplot as plt, koreanize_matplotlib, sys
sys.path.append(r'C:\Users\Win11Pro\Desktop\KDT-14\[7]_CVISION')
from cv_utils import printPolts
```

```python
# 데이터 준비
IMG_FILE = '../Data/images/girl.jpg'
```

```python
# 이미지 데이터 로딩 및 기본 정보 확인
img = cv2.imread(IMG_FILE)
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
print(f'[기본정보] shape : {img.shape}, ndim : {img.ndim}, dtype : {img.dtype}')
```

```python
# 마스크용 이미지 객체 생성 => 형태(shape) 동일
mask = np.zeros_like(img)

w, h = mask.shape[1], mask.shape[0]

# 추출/연산을 원하는 지점 1로 설정
# cv2.circle(mask, (150,140), 100, (255,255,255), -1)
cv2.circle(mask, (int(w/2), int(h/2)), 100, (255,255,255), -1)

# 값이 1인 부분만
masked = cv2.bitwise_and(img, mask)

plt.imshow(masked)
```

```python
img1 = cv2.imread('../Data/images/robot_arm1.jpg')
img2 = cv2.imread('../Data/images/robot_arm2.jpg')

img1 = cv2.cvtColor(img1, cv2.COLOR_BGR2GRAY)
img2 = cv2.cvtColor(img2, cv2.COLOR_BGR2GRAY)

diff = cv2.absdiff(img1, img2)
_, diff = cv2.threshold(diff, 1, 255, cv2.THRESH_BINARY)
diff_red = cv2.cvtColor(diff, cv2.COLOR_GRAY2BGR)
diff_red[:,:,2] = 0

spot = cv2.bitwise_xor(cv2.cvtColor(img2, cv2.COLOR_GRAY2BGR), diff_red)

printPolts(1,4, ['ads','ddd','qqq','www'], [img1,img2,diff,spot], ['gray'])
```
