### [ 이미지/영상 연산 ] <hr>
- 픽셀 단위 연산
- 픽셀 값 범위 : 0 ~ 255 => 부호없는 8bit int, uint8
    * 255를 넘는 수 => 오버플로우(overflow)
    * 0 보다 작은 수 => 언더플로우(underflow)

```python
import numpy as np, cv2, matplotlib.pyplot as plt, koreanize_matplotlib, sys
sys.path.append(r'C:\Users\Win11Pro\Desktop\KDT-14\[7]_CVISION')
from cv_utils import printPolts
```

```python
# 데이터 준비
IMG_FILE = '../Data/images/girl.jpg'
```

```python
# 이미지 데이터 로딩 및 기본 정보 확인
img = cv2.imread(IMG_FILE)
print(f'[기본정보] shape : {img.shape}, ndim : {img.ndim}, dtype : {img.dtype}')
```

```python
# 이미지 연산 => 밝기 조절
test = img.copy().astype(np.int16)
print(f'[기본정보] shape : {test.shape}, ndim : {test.ndim}, dtype : {test.dtype}')
```

```python
# + & - 연산
add = [test + value for value in range(10, 131, 40)]

for arr in add: arr[arr > 255] = 255

printPolts(1,5,['origin','+10','+50','+90','+120'], [img]+add)
```

```python
# + & - 연산
sub = [test - value for value in range(10, 131, 40)]
titles = [f'-{value}' for value in range(10, 131, 40)]

for arr in add: arr[arr < 0 ] = 0

printPolts(1,5,['origin']+titles, [img]+sub)
```
