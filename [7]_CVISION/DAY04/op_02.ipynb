```python
import numpy as np, cv2, matplotlib.pyplot as plt, koreanize_matplotlib, sys
sys.path.append(r'C:\Users\Win11Pro\Desktop\KDT-14\[7]_CVISION')
from cv_utils import printPolts
```

```python
# 데이터 준비
IMG_FILE1 = '../Data/images/man_face.jpg'
IMG_FILE2 = '../Data/images/lion_face.jpg'
```

```python
# 이미지 데이터 로딩 및 기본 정보 확인
man_img = cv2.imread(IMG_FILE1)
man_img = cv2.cvtColor(man_img, cv2.COLOR_BGR2RGB)
lion_img = cv2.imread(IMG_FILE2)
lion_img = cv2.cvtColor(lion_img, cv2.COLOR_BGR2RGB)
print(f'[기본정보] shape : {man_img.shape}, ndim : {man_img.ndim}, dtype : {man_img.dtype}')
print(f'[기본정보] shape : {lion_img.shape}, ndim : {lion_img.ndim}, dtype : {lion_img.dtype}')
```

```python
# 이미지 합성
ret = cv2.add(man_img, lion_img)
plt.imshow(ret);
```

```python
win_name = 'a'
def onChange(x):
    a = x/100
    dst = cv2.addWeighted(man_img, 1-a, lion_img, a, 0)
    cv2.imshow(win_name, dst)

cv2.imshow(win_name, man_img)
cv2.createTrackbar('fade', win_name, 0, 100, onChange)

cv2.waitKey()
cv2.destroyAllWindows()
```

```python
weight = np.linspace(0,1,6)

w = [cv2.addWeighted(lion_img, w, man_img, 1-w, 0) for w in weight]
title = [f'비율 {w:0.2f}:{1-w:0.2f}' for w in weight]

printPolts(1,6, title, w)
```
