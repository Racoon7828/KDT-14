### [ 객체 탐지 모델 - Haar Cascade ] <hr>
- openCV에서 제공
- 정면 얼굴, 상반신, 하반신, 눈동자, 웃는 얼굴 등 감지 모델
- 모델 파일 : xml 제공

```python
import numpy as np, cv2, matplotlib.pyplot as plt, koreanize_matplotlib
```

```python
IMG_FILE = '../Data/images/jeny.jpg'
# IMG_FILE = '../Data/images/yongju_01_cool.png'
MODEL_FILE = '../Data/Haar/haarcascade_frontalface_alt2.xml'
```

```python
# 정면 얼굴 감지 모델 객체 생성
detector = cv2.CascadeClassifier(MODEL_FILE)
```

```python
# 이미지/사진에서 얼굴 감지
imgNP = cv2.imread(IMG_FILE)

# 얼굴 감지
result = detector.detectMultiScale(imgNP)
print(f'감지 결과 : {len(result)}개, {result}')
```

```python
# 얼굴 부분에 사각형 그리기
x, y, w, h = result[0]
cv2.rectangle(imgNP, (x,y), (x+w, y+h), (255,0,0), 3)
cv2.putText(imgNP, "Bingo",(x+10, y - 10), cv2.FONT_HERSHEY_DUPLEX, 1, (255,0,0))
```

```python
# 화면에 뛰우기
cv2.imshow("DETECT", imgNP)
cv2.waitKey()
cv2.destroyWindow("DETECT")
```
