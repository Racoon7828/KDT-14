### [ 그리기 기능 ] <hr>
- 동영상/이미지 위에 객체 위치, 텍스트 출력 사용
- 기능 : 선그리기 / 사각형 그리기 / 원 그리기 / 타원 그리기 / 다각형 그리기 / 텍스트

```python
import numpy as np, cv2, matplotlib.pyplot as plt, koreanize_matplotlib
```

```python
# 그림 그릴 바탕 / 종이 생성
# 픽셀 값 : 0 ~ 255, 0을 포함한 양수 unsigned int8
w_paper = np.full((500,500, 3), 255, dtype=np.uint8)
```

```python
# 선그리기 : 좌표 2개 필요
sx, ex, sy, ey = 50,450,50,50

cv2.line(w_paper, (sx,sy), (ex,ey), (255,0,0)) # [default] 선 두께 - 1px, 선타입 - cv2.LINE_4
cv2.line(w_paper, (sx,sy+50), (ex,ey+50), (0,255,0), 10)
cv2.line(w_paper, (sx,sy+100), (ex,ey+100), (0,0,255), 20, cv2.LINE_4)
cv2.line(w_paper, (sx,sy+150), (ex,ey+150), (255,255,0), 20, cv2.LINE_8)
cv2.line(w_paper, (sx,sy+200), (ex,ey+200), (0,255,255), 20, cv2.LINE_AA) 
cv2.line(w_paper, (0,0), (500,500), (255,0,255))
```

```python
# 도화지 화면 출력
plt.imshow(w_paper);
```

```python
w_paper = np.full((500,500, 3), 255, dtype=np.uint8)
# 사각형 그리기 : 좌표 2개 필요, 너비 높이 자동 계산
cv2.rectangle(w_paper, (50,50), (150,150), (255,0,0))
cv2.rectangle(w_paper, (450,200), (200,450), (255,0,0), -1)

plt.imshow(w_paper)
```

```python
# 사각형 그리기 : 좌표 여러개 필요, 첫번째와 마지막 좌표 연결 여부
w_paper = np.full((500,500, 3), 255, dtype=np.uint8)

pts1 = np.array([[50,50], [150,150], [100,140], [200,240]], dtype=np.int32)
pts2 = np.array([[150,300], [50,450], [250,450], [250,450]], dtype=np.int32)
pts3 = np.array([[350,250], [450,350], [400,450], [300,450], [250,350]], dtype=np.int32)


cv2.polylines(w_paper, [pts1], False, (255,0,0))
cv2.polylines(w_paper, [pts2], True, (0,0,255), 10)
cv2.polylines(w_paper, [pts3], True, (0,0,0))

plt.imshow(w_paper);
```

```python
w_paper = np.full((500,500, 3), 255, dtype=np.uint8)

cv2.circle(w_paper, (150,150), 100, (255,0,0), -1)
cv2.ellipse(w_paper, (150,150), (100,100), 0,0, 180, (0,0,255), -1)
cv2.ellipse(w_paper, (100,150), (50,50), 0,0, 180, (255,0,0), -1)
cv2.ellipse(w_paper, (200,150), (50,50), 0,181, 360, (0,0,255), -1)

plt.imshow(w_paper);
```

```python
w_paper = np.full((500,500, 3), 255, dtype=np.uint8)
cv2.putText(w_paper, "Simplex", (50,30), cv2.FONT_HERSHEY_SIMPLEX, 1, (0,0,0))
cv2.putText(w_paper, "DUPLEX", (50,70), cv2.FONT_HERSHEY_DUPLEX, 1, (0,0,0))
cv2.putText(w_paper, "Complex", (50,220), cv2.FONT_HERSHEY_COMPLEX, 1, (0,0,0))
plt.imshow(w_paper);
```
