### [ 컬러 스페이스 - RGB계열 ] <hr>
- RGB/BGR : 3개의 색상 정보 조합한 색상 표현
- RGBA/RGRA : 4개의 색상 정보 조합한 색상 표현
    * A : 알파 채널. 투명도 수치값 저장. png, svg, gif, webp 투명 지정 가능한 포멧

```python
import numpy as np, cv2, matplotlib.pyplot as plt, koreanize_matplotlib, os
```

```python
IMG_FILE ='../Data/images/opencv_logo.png'
```

```python
img3 = cv2.imread(IMG_FILE)
img4 = cv2.imread(IMG_FILE, cv2.IMREAD_UNCHANGED)
```

```python
# 기본 정보 출력
print('--- img3 정보')
print(f'shape : {img3.shape}, ndim : {img3.ndim}')
print('--- img4 정보')
print(f'shape : {img4.shape}, ndim : {img4.ndim}')
```

```python
# 채널별 분리 화면 출력
fig, axes = plt.subplots(1, 3, sharey=True)
b, g, r = cv2.split(img3)

for i in range(3):
    axes[i].imshow([b,g,r][i], cmap='gray')
    axes[i].set_title(["B", "G", "R"][i])
    axes[i].axis('off')

# for ax, data, title in zip(axes, (b,g,r), ("B", "G", "R")):
#     ax.set_title(f'{title} Channel')
#     ax.imshow(data, cmap='gray')
```

```python
# 채널별 분리 화면 출력
fig, axes = plt.subplots(1, 4, sharey=True)
b, g, r, a = cv2.split(img4)

for i in range(4):
    axes[i].imshow([b,g,r,a][i])
    axes[i].set_title(["B", "G", "R", "A"][i])
    axes[i].axis('off')
```
