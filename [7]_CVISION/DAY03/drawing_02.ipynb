### [ 그리기 + 이벤트 처리 ] <hr>
- 이벤트
    * 키보드 입력 이벤트 => cv2.waitkey()
    * 마우스 입력 이벤트 => cv2.setMouseCallback() - 등록, onMouse() - 이벤트 처리

```python
import numpy as np, cv2, matplotlib.pyplot as plt, koreanize_matplotlib
```

```python
# 그림 그릴 바탕 / 종이 생성
# 픽셀 값 : 0 ~ 255, 0을 포함한 양수 unsigned int8
w_paper = np.full((500,500, 3), 255, dtype=np.uint8)
W_TITLE = 'Click!'
```

```python
# 마우스 이벤트 처리 함수 구현
# => onMouse()
# => event : cv2.EVENT_XXXX
# => x, y : 마우스 좌표값
# => flags : 미우스 이벤트와 함께 발생한 동작 cv2.EVENT_FLAG_XXXX
# => parms : 부가적으로 전달할 데이터들
def onMouse(event, x, y , flags, param):
    # print(f'event : {event}, ({x},{y})')

    if event == cv2.EVENT_LBUTTONDOWN:
        cv2.circle(w_paper, (x,y), 10, (0,255,0), -1)
        cv2.imshow(W_TITLE, w_paper)
```

```python
# 마우스 이벤트 처리 창 생성
cv2.imshow(W_TITLE, w_paper)
cv2.setMouseCallback(W_TITLE, onMouse)

while True:
    key = cv2.waitKey(1)        # 단일 waitKey로 통합
    if key in [ord('X'), ord('x')]:
        print('X')
        break
    
    if key in [ord('Q'), ord('q')]:
        w_paper = np.full((500, 500, 3), 255, dtype=np.uint8)
        cv2.imshow(W_TITLE, w_paper)  # 즉시 화면 갱신

cv2.destroyWindow(W_TITLE)
```
