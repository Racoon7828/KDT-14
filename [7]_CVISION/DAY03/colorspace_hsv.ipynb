### [ 컬러 스페이스 - HSV계열 ] <hr>
- HSV 색 공간
    * 특징 : H(색상) 채널이 조명 변화와 무관하게 안정적 -> 색깔로 물체 분리에 최적
    * HSI, HSL 이라고 함. (세부 계산 방식은 다름)
    * cv2.cvtColor( cv2.COLOR_BGR2HSV) 변환 필수

```python
import numpy as np, cv2, matplotlib.pyplot as plt, koreanize_matplotlib, os
```

```python
IMG_FILE ='../Data/images/hsv_fruits.jpg'
```

```python
## -------------------------------------------------------------
## 함수이름 : show_channels
## 함수기능 : 한 행에 원본 + 채널들을 표시하는 함수
## -------------------------------------------------------------
def show_channels(axes_row, channels, names, cmaps, original_img):
    axes_row[0].imshow(cv2.cvtColor(original_img, cv2.COLOR_BGR2RGB))
    axes_row[0].set_title("원본 이미지")
    for ax, ch, name, cmap in zip(axes_row[1:], channels, names, cmaps):
        ax.imshow(ch, cmap=cmap)
        ax.set_title(name)
```

```python
## ------------------------------------------------------------
## [실습 1] BGR vs HSV 채널 분리 비교
## ------------------------------------------------------------
# 이미지 로딩
imgNP = cv2.imread("../Data/Images/hsv_fruits.jpg")
imgHSV = cv2.cvtColor(imgNP, cv2.COLOR_BGR2HSV)

# BGR 채널 분리
b, g, r = cv2.split(imgNP) # BGR 채널 분리
# HSV 채널 분리
h, s, v = cv2.split(imgHSV) # HSV 채널 분리
```

```python
fig, axes = plt.subplots(2, 4, figsize=(16, 8))
fig.suptitle("BGR 채널 vs HSV 채널 비교", fontsize=15, fontweight="bold")
## BGR 원본과 BGR 채널 분리
show_channels(axes[0], [b, g, r], ["B 채널", "G 채널", "R 채널"], ["Blues", "Greens", "Reds"], imgNP)
## BGR 원본과 HSV 채널 분리
show_channels(axes[1], [h, s, v], ["H (색상)", "S (채도)", "V (명도)"], ["hsv", "YlOrBr", "gray"], imgNP)
plt.tight_layout()
plt.savefig("result_hsv_channels.jpg", dpi=120, bbox_inches="tight")
plt.show()
```
