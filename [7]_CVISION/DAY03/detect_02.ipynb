- [실습] 웃는 얼굴 감지 모델, 정면 얼굴 감지 모델 + 동영상/카메라

```python
import numpy as np, cv2, matplotlib.pyplot as plt, koreanize_matplotlib
```

```python
VID_FILE = '../Data/videos/person_01.mp4'
CAMERA_ID = 0
MODEL_FRONT = '../Data/Haar/haarcascade_frontalface_alt2.xml'
MODEL_SMILE = '../Data/Haar/haarcascade_smile.xml'
```

```python
# 얼굴 감지 모델 객체 생성
detector_front = cv2.CascadeClassifier(MODEL_FRONT)
detector_smile = cv2.CascadeClassifier(MODEL_SMILE)
```

```python
cap = cv2.VideoCapture(VID_FILE)
# cap = cv2.VideoCapture(CAMERA_ID)
cap.set(cv2.CAP_PROP_FRAME_WIDTH, 480)
cap.set(cv2.CAP_PROP_FRAME_HEIGHT, 360)

if cap.isOpened():
    fps = cap.get(cv2.CAP_PROP_FPS)
    delay = int(1000 / fps)

    while True:
        ret, frame = cap.read()
        if not ret:
            break
        
        # - minNeighbors : 지정된 숫자 만큼 감지된 결과물
        # - scaleFactor : 이미지/영상 크기 축소 비율값
        # - minsize : 감지 이미지 최소 크기
        # - maxsize : 감지 이미지 최대 크기
        res_front = detector_front.detectMultiScale(frame)
        # res_smile = detector_smile.detectMultiScale(frame)

        for (x, y, w, h) in res_front:
            cv2.rectangle(frame, (x, y), (x+w, y+h), (255, 0, 0), 3)
            cv2.putText(frame, "face", (x+25, y-10), cv2.FONT_HERSHEY_DUPLEX, 1, (255, 0, 0))

        rects, weights = detector_smile.detectMultiScale2(frame)

        for (x, y, w, h), weight in zip(rects, weights):
            if weight < 500:          # 3번 미만 검출된 건 무시 (값 올릴수록 엄격)
                continue
            cv2.rectangle(frame, (x, y), (x+w, y+h), (0, 255, 0), 3)
            cv2.putText(frame, f"smile({weight})", (x+10, y-10), cv2.FONT_HERSHEY_DUPLEX, 0.7, (0, 255, 0))

        # for (x, y, w, h) in res_smile:
        #     cv2.rectangle(frame, (x, y), (x+w, y+h), (0, 255, 0), 3)
        #     cv2.putText(frame, "smile", (x+25, y-10), cv2.FONT_HERSHEY_DUPLEX, 1, (0, 255, 0))

        cv2.imshow("DETECT", frame)
        if cv2.waitKey(delay) & 0xFF == ord('q'):
            break

cap.release()
cv2.destroyAllWindows()
```
