### [ IMAGE READ & WRITE ] <hr>
- 경로지정된 이미지 읽기 / 쓰기 즉, 저장
- 읽기/로딩 : cv2.imread(경로+파일명, 플래그)
- 쓰기/저장 : cv2.imwrite(저장_경로+파일명, 이미지 데이터)

```python
import cv2
import matplotlib.pyplot as plt, matplotlib.cm as cm, koreanize_matplotlib
```

```python
IMG_FILE =  '../Data/images/doge.jpg'
# IMG_FILE =  '../Data/images/19c9c4348c23a338.png'
```

```python
# 이미지 데이터 로딩 및 정보 확인
# - 함수 : cv2.imread()
# - 반환 : ndarray 이미지 데이터
color_img = cv2.imread(IMG_FILE) # BGR
print(f'{color_img.shape}, {color_img.dtype}, {color_img.size:,} bytes')
```

```python
# 이미지 데이터 로딩 및 정보 확인
# - 함수 : cv2.imwrite(저장경로+파일명, 이미지 데이터)
SAVE_FILE = './copy.jpg'

ret = cv2.imwrite(SAVE_FILE, color_img)
```

```python
# s 입력시 GRAY 이미지로 저장, x 종료
if color_img is not None:
    cv2.imshow('img', color_img)
    while True:
        if cv2.waitKey() in [ord('s'), ord('S')]:
            gray_img = cv2.imread(SAVE_FILE, cv2.IMREAD_GRAYSCALE)
            # gray_img = cv2.cvtColor(color_img, cv2.COLOR_BGR2GRAY)
            cv2.imwrite('output.jpg', gray_img)

        if cv2.waitKey() in [ord('x'), ord('X')]:
            cv2.destroyWindow('img')
            break
else: print("ㅗ")
```
