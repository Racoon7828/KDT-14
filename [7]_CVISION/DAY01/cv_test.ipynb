# 이미지를 3개 선택
- 1개 이미지 =>, 다른 1개 이미지 => 흑백, 마지막 1개 이미지 => 원본 그대로 읽기
- 3개 이미지 창 뛰우기
- 2초 간격으로 1번 이미지 => 2번 이미지 => 3번 이미지 닫기

```python
import cv2
```

```python
DIR_PATH = '../Data/images'
FILES = ['doge.jpg','19752893c8538724f.jpg','Nyan_cat_250px_frame.png']
FLAGS = [ cv2.IMREAD_COLOR, cv2.IMREAD_GRAYSCALE, cv2.IMREAD_UNCHANGED]
```

```python
# 3개 이미지 로딩 및 창 뛰우기
for fname, flag in zip(FILES, FLAGS):
    # 로딩할 이미지 경로 + 파일명
    img_path = f'{DIR_PATH}/{fname}'

    # 이미지 로딩
    imgNP = cv2.imread(img_path)
    # print(fname, flag)

    # 이미지 창 뛰우기
    cv2.imshow( fname, imgNP )

# 2초 간격으로 특정 이미지 창 닫기
for fname in FILES:
    # 이벤트 대기
    cv2.waitKey(2000)
    cv2.destroyWindow(fname)
```
