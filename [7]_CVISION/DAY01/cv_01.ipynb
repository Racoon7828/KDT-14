```python
import cv2
import matplotlib.pyplot as plt, matplotlib.cm as cm, koreanize_matplotlib
```

```python
# 주피터 패키지 설치 pip기반 => %pip install 패키지명
# %pip install koreanize_matplotlib
```

```python
IMG_FILE =  '../Data/images/inhwan_01_smile.png'
# IMG_FILE =  '../Data/images/19c9c4348c23a338.png'
```

```python
# 이미지 데이터 로딩 및 정보 확인
# - 함수 : cv2.imread()
# - 반환 : ndarray 이미지 데이터
img = cv2.imread(IMG_FILE) # BGR
print(f'{img.shape}, {img.dtype}, {img.size:,} bytes')
```

```python
# 이미지의 로우 데이터
print(f'{img.max()}, {img.min()}')
# print(img)
```

```python
# 이미지 로우 데이터 => 화면 출력
# => openCV cosjf tnstj : BGR
img = img[..., [2,1,0]] # RGB 변한
plt.imshow(img);
```

```python
# 이미지 데이터 로딩 및 정보 확인
# - 함수 : cv2.imread()
# - 옵션/플래그 : cv2.IMGGREAD_GRAYSCALE
# - 반환 : ndarray 이미지 데이터
img = cv2.imread(IMG_FILE, cv2.IMREAD_GRAYSCALE) # 흑백
```

```python
# 이미지 로우 데이터 => 화면 출력
# => openCV 채널 순서 : BGR
# => matplotlib 설정된 컬러맵 vivid => 변경
plt.imshow(img, cmap=cm.gray);
plt.title(f"형태: {img.shape}")
```

```python
# 이미지 담은 창 뛰우기 : cv2.imshow(창_타이틀, 이미지 데이터)
# (1) 창 생성 및 이미지 로딩
cv2.imshow('img', img)

# (2)
cv2.waitKey(delay=1000)

# (3) 키 입력 시 창 닫기
cv2.destroyAllWindows()
# cv2.destroyWindow("img")
```

```python
# 이미지 담은 창 뛰우기 : cv2.imshow(창_타이틀, 이미지 데이터)
# 'q' 입력 시 창 닫기
if img is not None:
    # print(img.size)
    cv2.imshow('img', img)

    while True:
        if cv2.waitKey() in [ord('q'), ord('Q')]:
            cv2.destroyWindow('img')
            break
else:
    print(img)
```
