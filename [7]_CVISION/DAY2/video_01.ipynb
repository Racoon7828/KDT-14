<### [ 동영상 & 카메라 영상 제어 ] <hr>
- 영상 : 이미지를 초당 27게 ~ 40개 보여주는 것
- 단위 : FPS (Frames Per Second)
- 전용 클래스 : VideoCapture
- 필수 메서드
    * 연결 ㅁㅊ 초기화 여부 체크 : isOpened() -> True/False
    * 프레임 읽어오기/로딩하기 : read() -> True/False, ndarray image data
    * 연결 및 자원 해제 반납 : release()

```python
import cv2
import os
```

```python
# 비디오
video_path = '../Data/videos/Media_pc35ohhKITs_001_720p.mp4'
# video_path = '../Data/videos/Astartes-II-Official-Teaser-Trailer-Warh_Media_sbgYaeerXhg_001_1080p.mp4'
CAM_ID = 0 # 컴퓨터 카메라 번호
F_NAME = 'crayon' # 프레임 저장시 파일명

# 폴더 생성
os.makedirs(F_NAME, exist_ok=True) # 프레임 저장용 폴더 생성
```

```python
cap = cv2.VideoCapture(video_path)

if cap.isOpened():

    cnt = 0
    fps = cap.get(cv2.CAP_PROP_FPS)
    delay = int(1000 / fps)

    print(f"FPS : {fps} Delay : {delay}")

    while True:
        ret, frame = cap.read()   # 프레임 한 장 읽기
        
        if not ret:               # 영상 끝나면 종료
            break
        
        cv2.imshow('video', frame)
        cv2.imwrite(f'./{F_NAME}/{F_NAME}_{cnt:0>3}.jpg', frame)
        cnt += 1

        # if cv2.waitKey(25) & 0xFF == ord('q'):
        if cv2.waitKey(delay) & 0xFF == ord('q'):
            break
else: print("ㅗ")

cap.release()
cv2.destroyAllWindows()
```
