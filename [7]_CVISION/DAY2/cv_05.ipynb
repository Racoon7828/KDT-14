### [ 이미지 크기 변경 ] <hr>
- 가장 많이/필수로 하는 전처리 과정
- 이미지마다 크기가 다르기 때문에 모든 이미지 크기 일치를 위해서 사용됨!
- 종류
    * 절대 크기 변경 : 이미지들의 크기를 고려하지 않고 동일 크기 일치 시킴
    * 상대 크기 변경 : 현재 이미지의 크기를 기준으로 조정. 이미지 마다크기 다름
- 함수 : cv2.resize()

```python
import cv2
import matplotlib.pyplot as plt, koreanize_matplotlib
import matplotlib.cm as cm
```

```python
# 데이터 준비
FILE_PATH = '../Data/images/doge.jpg'
```

```python
# 이미지 데이터 로딩
imgNP = cv2.imread(FILE_PATH)
```

```python
# 이미지 절대 크기 조정
imgAbs = cv2.resize(imgNP, (100,100))
print(f'shape : {imgAbs.shape}, ndim : {imgAbs.ndim}D')
```

```python
# 이미지 상대 크기 조정
imgRel = cv2.resize(imgNP, (0,0), fx=1.5, fy=0.5)

print(f'shape : {imgRel.shape}, ndim : {imgRel.ndim}D')
```

```python
## 원본 이미지, 절대 , 상대
fig, axes = plt.subplots(1,3)
for i in range(3):
    axes[i].imshow([imgNP, imgAbs, imgRel][i])
    axes[i].set_title(['imgNP', 'imgAbs', 'imgRel'][i])
    axes[i].axis('off')
plt.show()
```
