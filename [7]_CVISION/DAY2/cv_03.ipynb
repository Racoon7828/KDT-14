### [ 이미지 데이터 채널 분리 & 병합 ] <hr>
- 채널 분리 : cv2.split()
    - openCV BGR 순서의 채널
- 채널 병합 : cv2.merge((채널, 채널, 채널))    

```python
import cv2
import matplotlib.pyplot as plt, koreanize_matplotlib
import matplotlib.cm as cm
```

```python
# 데이터 준비
FILE_PATH = '../Data/images/doge.jpg'
```

```python
# 이미지 데이터 로딩
imgNP = cv2.imread(FILE_PATH)

print(f'shape : {imgNP.shape}, ndim : {imgNP.ndim}D')
```

```python
# openCV & matplotlib
plt.imshow(imgNP) # BGR
plt.show()

cv2.imshow('doge', imgNP)
cv2.waitKey()
cv2.destroyWindow('doge')
```

```python
# 채널 분리 : cv.split()
b_ch, g_ch, r_ch = cv2.split(imgNP)

print(f'BGR => {b_ch.shape}, {g_ch.shape}, {r_ch.shape}')

# 채널별로 시각화 : 1행 3열
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

for i in range(3):
    axes[i].imshow([b_ch, g_ch, r_ch][i], cmap=cm.gray)
    axes[i].set_title(['Blue', 'Green', 'Red'][i])
    axes[i].axis('off') # 축 정보 숨기기

plt.show()
```

```python
# 채널 병합 : cv2.merge((채널1,...)) : 원하는 순서대로
# BGR ndarray ==> RGB ndarray
rgbNP = cv2.merge((r_ch, g_ch, b_ch))

fig, axes = plt.subplots(1,2)

titles = ['imgNP', 'rgbNP']
datas = [imgNP, rgbNP]

for ax, title, data in zip(axes, titles, datas):
    ax.set_title(title)
    ax.imshow(data)
    ax.axis('off') # 축 정보 숨기기

plt.show()
```
