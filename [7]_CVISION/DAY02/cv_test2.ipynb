```python
import cv2
```

```python
FILE_PATH = '../Data/images/doge.jpg'
imgNP = cv2.imread(FILE_PATH)
imgNP = cv2.cvtColor(imgNP, cv2.COLOR_BGR2RGB)
```

```python
# 이미지 담은 창 뛰우기 : cv2.imshow(창_타이틀, 이미지 데이터)
# 'u' 누를 때마다 이미지 커짐, 'd' 누를 때마다 이미지 작아짐, 'x' 누르면 창 닫기

if imgNP is not None:
    scale = 1.0
    cv2.imshow('img', imgNP)

    while True:
        key = cv2.waitKey(0)

        if key == ord('+'):
            scale += 0.1
            h, w = imgNP.shape[:2]
            resized = cv2.resize(imgNP, (int(w * scale), int(h * scale)))
            cv2.imshow('img', resized)

        elif key == ord('-'):
            scale = max(0.1, scale - 0.1)
            h, w = imgNP.shape[:2]
            resized = cv2.resize(imgNP, (int(w * scale), int(h * scale)))
            cv2.imshow('img', resized)

        elif key in [ord('X'), ord('x')]:
            cv2.destroyWindow('img')
            break
else:
    print(imgNP)
```
