<### [ 동영상 & 카메라 영상 제어 ] <hr>
- VideoCapture의 속성들
- 속성명 형식 : cv2.CAP_PROP_속성명
- OpenCV에서 미리 전부 선언 / 정의해둠

- 필수 메서드
    * 현재 속성값 일기 : get(속성ID)
    * 현재 속성값 변경 : set(속성ID , 새로운 값)

```python
import cv2
```

```python
# 비디오
video_path = '../Data/videos/Media_pc35ohhKITs_001_720p.mp4'
# video_path = '../Data/videos/Astartes-II-Official-Teaser-Trailer-Warh_Media_sbgYaeerXhg_001_1080p.mp4'
```

```python
# 객체 생성
cap = cv2.VideoCapture(video_path)

# 현재 속성 정보 출력
print(f'{cap.get(cv2.CAP_PROP_FORMAT)}')       # 포맷
print(f'{cap.get(cv2.CAP_PROP_FOURCC)}')       # 4개의 캐릭터 : 코텍, 정보표시
print(f'{cap.get(cv2.CAP_PROP_FPS)}')          # 1초당 프레임 수
print(f'{cap.get(cv2.CAP_PROP_FRAME_HEIGHT)}') # 프레임 높이/행
print(f'{cap.get(cv2.CAP_PROP_FRAME_WIDTH)}')  # 프레임 너비/열
print(f'{cap.get(cv2.CAP_PROP_FRAME_COUNT)}')  # 프레임 개수
```

```python
# 컴퓨터 연결된 카메라 로딩
CAMERA_ID = 0
cap = cv2.VideoCapture(CAMERA_ID)
print(f'{cap.isOpened()}')

print(f'{cap.get(cv2.CAP_PROP_FORMAT)}')       # 포맷
print(f'{cap.get(cv2.CAP_PROP_FOURCC)}')       # 4개의 캐릭터 : 코텍, 정보표시
print(f'{cap.get(cv2.CAP_PROP_FPS)}')          # 1초당 프레임 수
print(f'{cap.get(cv2.CAP_PROP_FRAME_HEIGHT)}') # 프레임 높이/행
print(f'{cap.get(cv2.CAP_PROP_FRAME_WIDTH)}')  # 프레임 너비/열
print(f'{cap.get(cv2.CAP_PROP_FRAME_COUNT)}')  # 프레임 개수

cap.set(cv2.CAP_PROP_FRAME_WIDTH, 300)
cap.set(cv2.CAP_PROP_FRAME_HEIGHT, 200)

print(f'{cap.get(cv2.CAP_PROP_FRAME_WIDTH)}')  # 프레임 너비/열
print(f'{cap.get(cv2.CAP_PROP_FRAME_HEIGHT)}') # 프레임 높이/행
```

```python
if cap.isOpened:
    while True:
        ret, frame = cap.read()

        if not ret: break 
        cv2.imshow('CAMERA', frame)
        
        if cv2.waitKey(30) & 0xFF == ord('q'):
            break

cap.release()
cv2.destroyAllWindows()
```

```python
if cap.isOpened():
    fps = cap.get(cv2.CAP_PROP_FPS)
    delay = int(1000 / fps)

    print(f"FPS : {fps} Delay : {delay}")

    while True:
        ret, frame = cap.read()   # 프레임 한 장 읽기
        
        if not ret: break              # 영상 끝나면 종료
        
        cv2.imshow('video', frame)
        if cv2.waitKey(delay) & 0xFF == ord('q'):
            break
else: print("ㅗ")

cap.release()
cv2.destroyAllWindows()
```
