### [ 이미지 색공간 변경 ] <hr>
- 색 공간 : 이미지 색상을 표현하는 다양한 방식들. 목적에 따라 적합한 색공간 사용 필요
- 색 송간 변환 함수 : cv2.cvtColor()
- 색 공간 변환 상수 : cv2.COLOR_BRG2HSV,... => cv2.COLOR_XXX2XXX

```python
import cv2
import matplotlib.pyplot as plt, koreanize_matplotlib
import matplotlib.cm as cm
```

```python
# 데이터 준비
FILE_PATH = '../Data/images/doge.jpg'
```

```python
# 이미지 데이터 로딩
imgNP = cv2.imread(FILE_PATH)
```

```python
# 색공간 변경 : cv2.ctrColor(imgNP, cv2.COLOR_XXXX)
rgb_img = cv2.cvtColor(imgNP, cv2.COLOR_BGR2RGB)
gray_img = cv2.cvtColor(imgNP, cv2.COLOR_BGR2GRAY)

fig, axes = plt.subplots(1,3, figsize=(15, 5))
for i in range(3):
    axes[i].imshow([imgNP, rgb_img, gray_img][i], cmap=cm.gray)
    axes[i].set_title(['imgNP', 'rgb_img', 'gray_img'][i])
    axes[i].axis('off') # 축 정보 숨기기

plt.show()
```
