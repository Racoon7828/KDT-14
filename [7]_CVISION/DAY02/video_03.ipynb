### [ 동영상 & 카메라 영상 제어 ] <hr>
- VideoWriter 동영상 저장 클래스
- 동영상 포맷 형식, 초당 프레임 수, 프레임 크기 설정

- VideoWriter 객체 생성
- 필수 메서드
    - fourcc(c1, c2, c3, c4) : 포맷 방식 문자 설정 => cv2.VideoWriter_fourcc() 함수
    - write(ndaray 이미지) : 비디오 프레임으로 저장
    - release(): VideoWriter 객체 생성

```python
import cv2
```

```python
# 연결 카메라 장치 번호
CAMERA_ID = 0
```

```python
# 카메라 로딩
cap = cv2.VideoCapture(CAMERA_ID)

# 현재 속성 변경
cap.set(cv2.CAP_PROP_FRAME_WIDTH, 300)
cap.set(cv2.CAP_PROP_FRAME_HEIGHT, 200)

print(f'{cap.get(cv2.CAP_PROP_FRAME_WIDTH)}')  # 프레임 너비/열
print(f'{cap.get(cv2.CAP_PROP_FRAME_HEIGHT)}') # 프레임 높이/행
```

```python
# 카메라 영상 저장/녹회
# 녹화 관련 설정
save_video = './crayon/save.avi' # 녹화 파일 경로
fps = 25.40                      # 초당 프레임 수

# 프레임 사이즈 정보
ssize = (int(cap.get(cv2.CAP_PROP_FRAME_WIDTH)), int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))) 

# 포맷 정보
# fourcc = cv2.VideoWriter_fourcc('D','I','V','X')
fourcc = cv2.VideoWriter_fourcc(*'DIVX')

if cap.isOpened:
    # 영상 녹화 객체
    record = cv2.VideoWriter(save_video, fourcc, fps, ssize)

    while True:
        ret, frame = cap.read()

        if not ret: break 
        cv2.imshow('CAMERA', frame)
        record.write(frame)
        
        if cv2.waitKey(1) & 0xFF == ord('q'):
            break

cap.release()
cv2.destroyAllWindows()
```
