### [ Dataset & DataLoader : 데이터 전담 클래스 ]
- 목적 : 대량 데이터의 효율적 처리와 관리를 위한 pytorch 설계
- 모듈 : torch.utills.data.Dataset, DataLoader
    * Dataset : 추상 클래스. 상속받아서 자식 클래스 정의 필수
    * DataLoader : Dataset에서 지정된 배치크기 만큼 Tensor타입의 데이터와 타겟을 반환

```python
from torch.utils.data import Dataset, TensorDataset
import pandas as pd, numpy as np, torch
```

```python
# 데이터 준비 및 확인
df = pd.read_csv('../../[8] 머신러닝/Data/Numbers/iris.csv', usecols=[0,1,2,3])
df.head(2)
```

```python
# 피쳐와 타겟 분리
feature = df[df.columns[:-1]].values
target = df[df.columns[-1]].values

print(feature.shape, target.shape)
```

```python
# 피쳐와 타겟 Numpy -> tensor
feature = torch.tensor(feature)
target = torch.tensor(target)

print(feature.shape, target.shape)
```

> 기존 TensorDataset 클래스 활용

```python
# 전용 Dataset 인스턴스 생성 : Dataset의 자식 클래스 TensorDataset
#                           Map_style Dataset
ds = TensorDataset(feature, target)
```

> 사용자 정의 Dataset 클래스

```python
# -----------------------------------------------
# 데이터 : iris.csv
# 피 쳐 : 꽃받침 길이, 너비, 꽃잎 길이
# 타 겟 : 꽃잎 너비
# -----------------------------------------------
# 클래스 이름 : IrisDataset
# 부모 클래스 : Dataset
# 클래스 속성 : feature, target
# 클래스 기능 : 필수 오버라이딩 메서드
#              df __init__()
#              df __len__()
#              df __getitem__()
# -----------------------------------------------
class IrisDataset(Dataset):
    def __init__(self, feature, target):
        super().__init__()
        self.feature = feature.values # DF -> ndarray
        self.target = target.values # DF = ndarry
        self.cols = feature.columns # 컬럼이름 
        self.rows = feature.shape[0] # 샘플 수
    
    # 샘플 수 반환 메서드
    def __len__(self):
        return self.rows

    # 인덱스에 해당하는 피쳐와 타겟 텐서 반환 메서드
    def __getitem__(self, index):
        feature = torch.FloatTensor(self.feature[index])
        target = torch.LongTensor(self.target[index])
        return feature, target
```
