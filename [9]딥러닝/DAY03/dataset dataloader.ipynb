### [ Dataset & DataLoader : 데이터 전담 클래스 ]
- 목적 : 대량 데이터의 효율적 처리와 관리를 위한 pytorch 설계
- 모듈 : torch.utills.data.Dataset, DataLoader
    * Dataset : 추상 클래스. 상속받아서 자식 클래스 정의 필수
    * DataLoader : Dataset에서 지정된 배치크기 만큼 Tensor타입의 데이터와 타겟을 반환

```python
from torch.utils.data import Dataset, TensorDataset
import pandas as pd, numpy as np, torch, sys
sys.path.append('./custom_dataset.py')
from custom_dataset import *
from torch.utils.data import DataLoader
```

```python
# 데이터 준비 및 확인
df = pd.read_csv('../../[8] 머신러닝/Data/Numbers/iris.csv', usecols=[0,1,2,3])
df.head(2)
```

> 사용자정의 Dataset 클레스 + DataLoader 클래스

```python
irisDs = IrisDataset(df[df.columns[:-1]], df[df.columns[-1:]])
```

```python
# DataLoader 인스턴스 생성 : 학습에 사용될 배치 크기 만큼의 데이터, 타겟 로딩
bs = 60
irisDL = DataLoader(irisDs, batch_size=bs, shuffle=True)
```

```python
for idx, (data, target) in enumerate(irisDL, start=1):
    print(f'\n[{idx}]====={len(data)}개')
    print(data.shape, target.long().shape)
```
