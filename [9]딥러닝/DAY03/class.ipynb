#### [ 사용자 정의 클래스 ]
- 데이터 전용 모델, 데이터 전담 클래스 정의 및 생성
- 사용자 정의 클래스 정의 필요!
- 상속
- 필수 메서드 오버라이딩

```python
# [실습 : 서람과 학생 ]
# - 상속 체크 : 사람이 학생이다 (x), 학생이 사람이다 (O) : 부모 자식 관계
# - 포함 체크 : 사람들은 학생을 갖고 있다. (O), 학생들은 사람을 갖고 있다. (X)
# -----------------------------------------------------------------------------------
# 클래스 이름 : Person
# 클래스 역활 : 사람을 표현할 수 있는 클래스
# 클래스 속성 : 성별, 키, 몸무게, 이름, 나이
# 클래스 기능 : 먹는다, 잔다.
# -----------------------------------------------------------------------------------
class Person:
    loc = "대한민국"
    def __init__(self, name, age=0, height=0, weight=0, gender=None):
        self.name = name
        self.age = age
        self.height = height
        self.weight = weight
        self.gender = gender

    def sleep(self, where):
        print(f'{self.name}은 {where}에서 잔다')

    def eat(self, where, eat):
        print(f'{self.name}은 {where}에서 {eat}먹는다')

    @classmethod
    def print_info(cls):
        print(f'국적 : {cls.loc} 입니다')
        print('클래스 메서드로 인스턴스 속성/메서드 사용불가!')

    # 오버라이딩 메서드
    # - 함수 역활 : 인스턴스 변수명으로 호출 가능
    def __call__(self, data=None):
        print('calling', data)
        self.sleep('안방')

class Student(Person):
    school_name = '대구중학교'
    def __init__(self, name, grade):
        # 상속시 부모 클래스 부터 생성 => super
        super().__init__(name)
        self.grade = grade
        print(f'{self.name}은 학생이다')

    def study(self, subject):
        print(f'{self.name}은 {subject} 과목을 공부한다')

    def test(self, subject):
        print(f'{self.name}은 {subject} 과목을 시험친다')

    def sleep(self, where):
        print(f'{self.name}은 {where}에서 잘것이다')
```

```python
# 인스턴스/객체 생성 : 클래스명(매개변수) -> __init__(매개변수)
#                    생성자(Constructor)
hong = Person('Hong', 10, 17, 80, 'M')
hong.sleep('집')

park = Student('Park', 1)
park.eat('집','피자')
```

```python
print(hong.loc, park.loc)

Person.print_info()

hong.print_info()

park.print_info()
```
