#### [ Pytorch Tensor & Numpy ndarray ]
- 다차원 배열 데이터를 저장
- 공통 속성 : shape, ndim, dtype, ...
- Tensor => DL 모델이 입력 / 출력 형태
- ndarray => Pandas, matplotlib, scikit-learn, pytorch에서 데이터 저장 형태
- 조작 방식 비슷

```python
import torch
import numpy as np
```

```python
np = np.array([[11,22,33]])
tsr = torch.tensor([[66,77,88]])
```

```python
print(f'[ndarray 속성들]-----')
print(f'shape: {np.shape} ndim: {np.ndim} dtype: {np.dtype}')
print(f'shape: {tsr.shape} ndim: {tsr.ndim} dtype: {tsr.dtype}')
```
