### [Full Connected Layer - Linear Layer]
- 모든 뉴런들이 다음 층의 뉴런들과 연결되는 중
- Pytorch에서는 Linear 클래스로 규현되어 있음
- 피쳐와 가중치 곱셈 + 절현 연산 수행

- torch.nn.Linear 클래스
    * 주요 매개변수 : 입력 개수와 출력 개수
    * 해당 총에 퍼셉트론 생성 <= 출력 개수>
    * w, B 텐서 생성 및 초가화(랜덤)
    * 순전파 진행 후 결과 반환

```python
import torch
import torch.nn as nn # 인공신경망 관련 모듈
```

>> Linear 클래스 이해하기

```python
# 입력과 출력 선정
in_, out_ = 5, 10

# Linear 인스턴스 생성
# => w, b 텐서 생성 및 초기화(랜덤)
llayer = nn.Linear(in_, out_)
```

```python
print(llayer.weight)
print(llayer.bias)
```

```python
# 순전파(Forward) 진행 : 변수명( 입력 데이터 )
# data = torch.tensor([[11,22,33,44,55]], dtype=torch.float32)
data = torch.FloatTensor([[11,22,33,44,55]])
llayer(data)
```
