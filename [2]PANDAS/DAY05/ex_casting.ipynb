- DataFrame/Series의 원소/요소의 실제 데이터에 맞게 타입 변환 
    * 현재 타입 확인 속성 : dtype/dtypes
    * 현재 타입 변경 메서드 : 변수명.astype(변환타입명)반드시 저장 필수

````python
# 모듈 로딩
import pandas as pd # 데이터 분석용 패키지
import numpy as np  # 배열을 파이썬언어로 규현한 패키지
                    # 판타스에서 데이터 저장을 numpy 배열로 함
````

````python
## 데이터 준비 및 DF 인스턴스 생성
df = pd.DataFrame({
"이름": ["김민수", "이서연", "박지훈"],
"나이": ["20", "21", "19"],
"점수": ["85.5", "90.0", "78.5"],
"가입일": ["2024-01-10", "2024-02-15", "2024-03-01"]})
````

````python
df
````


- 방법1) DataFrame/Series.astype(변환_타입)

````python
## DF인스턴스 요약 정보
df.info()
````

````python
# => 실제 데이터와 타입이 맞지 않는 컬럼 확인
# => 나이 > int, 점수 > float, 가입일 > datetime64[ns]
# => 변환 자료형 : 타입명 입력 시 가장 큰 값의 타입
df["나이"] = df["나이"].astype("int8") # inplace 매개변수 없음, 반드시 저장
df["점수"] = df["점수"].astype('float32')
df["가입일"] = df["가입일"].astype("datetime64[ns]") # 날짜/시간 타입명은 반드시 문자열!!
df.dtypes
````

##### 여러 컬럼의 자료형을 한 번에 처리 후 저장
```python
df = df.astype({"나이": int,
                "점수": float,
                "가입일": "datetime64[ns]"})
```

````python
# => 변환 자료형 : 타입명 입력 시 가장 큰 값의 타입
# => 결측치 처리 가능한 타입 : 첫 글자 대문자
df["나이"] = df["나이"].astype("Int8") # inplace 매개변수 없음, 반드시 저장
df["점수"] = df["점수"].astype('Float32')
df["가입일"] = df["가입일"].astype("datetime64[ns]") # 날짜/시간 타입명은 반드시 문자열!! 권장 X
df.dtypes
````

- 방법2) 숫자형으로 변환 함수 - pands.to_numeric(데이터)

````python
## 데이터 준비 및 DF 인스턴스 생성
df = pd.DataFrame({
"이름": ["김민수", "이서연", "박지훈"],
"나이": ["20", "21", "19"],
"점수": ["85.5", "90.0", "78.5"],
"가입일": ["2024-01-10", "2024-02-15", "2024-03-01"]})
````

````python
## 나이, 점수 컬럼 object ==> 숫자 정수, 실수 형변환
print("변환 전 ============\n",df.dtypes)
````

````python
# 실수 기본 float64
df['점수'] = pd.to_numeric(df['점수'], downcast='float')
df['나이'] = pd.to_numeric(df['나이'], downcast='integer')
````

````python
print("변환 후 ============\n",df.dtypes)
````

- 방법3) 날짜시간 데이터 형변환 - pandas.to_datetime()

````python
# 가입일 컬럼 object >>> datetime64[ns] 형변환
df.가입일 = pd.to_datetime(df.가입일, dayfirst=True)
````

````python
print("변환 후 ============\n",df.dtypes)
````
