### [데이터 전처리를 위한 인코딩]
* 컬럼 중 문자형/범주형인 데이터를 숫자로 변환하는 작업
* 컴퓨터가 분석 및 학습을 하기 쉽도록 변환
* 종류
    * 정수 인코딩/ 정수 라벨링
    * 원-핫 인코딩

```python
# 모듈 로딩
import pandas as pd # 데이터 분석용 패키지
import numpy as np  # 배열을 파이썬언어로 규현한 패키지
                    # 판타스에서 데이터 저장을 numpy 배열로 함
import sys
sys.path.append('../utils')
from utils import data_info
```

```python
## 데이터 준비 및 DF 인스턴스 생성
DATA_FILE = '../DATA/iris.csv'

# 첫번째 줄 : 컬럼명 ok, 구분자 : 쉼표/콤마 ok
iris_df = pd.read_csv(DATA_FILE)
```

- 방법1) variety 컬럼 object ==> 숫자/정수 라벨 인코딩

```python
# 품종명에 설정될 값 매핑
irisDF_map = {
"Setosa": 0,
"Versicolor": 1,
"Virginica": 2}

## 결측값 NaN에는 함수 적용하여 NaN. 결과를 새로운 컬럼 추가
iris_df["variety_label"] = iris_df['variety'].map(irisDF_map)

iris_df[1::50]
```

- 방법2) 특정 값만 1로 설정하는 인코딩 : 원-핫 인코딩

```python
iris_df2 = pd.get_dummies(iris_df, columns=["variety"], dtype=int)
iris_df2[1::50]
```
