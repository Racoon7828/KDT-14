```python
import pandas as pd
import requests

# 1. 대상 URL 설정
url = 'https://ko.wikipedia.org/wiki/%EC%98%AC%EB%A6%BC%ED%94%BD_%EB%A9%94%EB%8B%AC_%EC%A7%91%EA%B3%84'

# 2. 브라우저처럼 보이게 헤더 설정 (중요)
headers = {'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/119.0.0.0 Safari/537.36'}

# 3. requests로 HTML 내용 가져오기
response = requests.get(url, headers=headers)

# 4. 가져온 HTML 텍스트를 pandas의 read_html에 전달
# (참고: pd.read_html은 리스트를 반환하므로 인덱싱이 필요함)
df_raw = pd.read_html(response.text)

# 첫 번째 테이블 확인
df_raw
```
