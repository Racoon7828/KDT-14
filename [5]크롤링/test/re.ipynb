### 정규 표현식
- 문자열의 패턴을 표현하는 특별한 문법
- 코드가 간결해지는 효용성
- 문자열의 규칙이 복작할수록 효용이 더 커진다

```python
# 정규표현식을 사용하여 주민등록증 뒷자리를 *로 대체하는 예시 
import re

data = """
park 800905-1049118
kim  700905-1059119
"""

pat = re.compile(r"(\d{6})-\d{7}")
print(pat.sub("\g<1>-*******", data))
```

```python
a = 'a before dude abcdefg'

# 문자열 a 중 a,b,c를 찾기
re.findall('[abc]', a)
```

```python
# 문자열 사이 하이픈(-) 두 문자 사이의 범위 지정
re.findall('[a-g]+', a)
```

```python
# 문자 클래스 맨 앞에 ^를 사용하면 해당 문자들을 제외
re.findall('[^a-f]+', a)
```

#### 자주 사용하는 범위 표현
- [a-zA-Z] : 모든 알파벳 (소문자와 대문자 모두)
- [0-9] : 모든 숫자 (0부터 9까지)
- [가-힣] : 모든 한글 (가부터 힣까지)

#### 자주 사용하는 문자 클래스
- \d - 숫자와 매치된다. [0-9]와 동일한 표현식이다.
- \D - 숫자가 아닌 것과 매치된다. [^0-9]와 동일한 표현식이다.
- \s - 화이트스페이스(whitespace) 문자와 매치된다. [ \t\n\r\f\v]와 동일한 표현식이다. 맨 앞의 빈칸은 공백 문자(space)를 의미
- \S - 화이트스페이스 문자가 아닌 것과 매치된다. [^ \t\n\r\f\v]와 동일한 표현식이다.
- \w - 문자+숫자(alphanumeric)와 매치된다. [a-zA-Z0-9_]와 동일한 표현식이다.
- \W - 문자+숫자(alphanumeric)가 아닌 문자와 매치된다. [^a-zA-Z0-9_]와 동일한 표현식이다.

```python
# 중간에 줄바꿈(\n)이 포함된 문자열
text = "hello\nworld"

# 1. re.DOTALL을 안 썼을 때 (기본값)
# 마침표(.)가 줄바꿈(\n)을 만나면 매칭을 중단하므로, 'hello'만 찾고 멈춥니다.
print(re.findall('hello.', text)) 
# 출력 결과: []  (hello 뒤에 줄바꿈이 와서 매칭 실패)


# 2. re.DOTALL을 썼을 때
# 마침표(.)가 줄바꿈(\n)까지 문자로 인정하고 삼켜버립니다.
print(re.findall('hello.', text, re.DOTALL))
# 출력 결과: ['hello\n'] (줄바꿈까지 포함해서 찾기 성공)
```

```python
# * 메타 문자 =>  반복을 의미
# * 바로 앞에 있는 문자 a가 0부터 무한대까지 반복될 수 있다
```

```python
# 메타 문자 +
# 최소 1번 이상 반복될 때 사용한다.
```

```python
# 정규 표현식
a = '이제 우리는 disco 춤을 춰 pang pang dance'
re.findall('[a-zA-Z가-힣]{4,5}', a)
```

```python
# 정성적인 정규 표현식 사용법
p = re.compile(('[a-zA-Z가-힣]{4,5}'))
p.findall(a)
```
